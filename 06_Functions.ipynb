```python
#Function--Excecutes when called 
#Four types of functions given below:-

#1.Function without Arguments& without return
def greet():
    print("Hello Welcome to python Programming..")
greet()

#2.Function with arguments & without return
def sum(a,b):
    print("Sum:",a+b)
sum(2,5)   

#3.Function without argumets and with return
def number():
    return 5
number()

#4.Function with arguments & with return
def square(n):
    return n*n
result=square(5)
print("Area of square:",result)
```

```python
#Printing odd and even
def num(n):
    n=int(input("Enter the number:"))
    if n % 2==0:
        print("Even")
    else:
        print("Odd")
num(n)            
```

```python
#Factorial using functions
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)
print(fact(6))    
```

```python
#Prime numbers
def prime(n):
    n=int(input("Enter a number:"))
    if n<=0:
        return "Not Prime"
    if n%2==0:
        return "Not Prime"
    else:
        print("Prime")
prime(n)        
```

```python
#Function with Default Arguments
def greet(name="Santhosh"):
    print("Hello",name)
greet("Python")
greet()
    
```

```python
#Function with multiple return values
def operations(a,b):
    return a+b,a-b,a*b
s,d,m=operations(10,5)
print("Sum:",s)
print("Sub:",d)
print("Multi:",m)
```

```python
#Second largest number
def second_largest(lst):
    first=second=-9999
    for i in lst:
        if i > first:
            second =first
            first=i
        elif i > second and i!=first:
            second=i
    print("Second Largest number",second) 
second_largest([10,24,90,406,605])                
```

```python
#fiboacci series
def series(n):
    a,b=0,1
    for i in range(n):
        print(a,end=" ")
        a,b=b,a+b
series(11)        
```

```python
#Palindrome checking
def palindrome(n):
    m=n[::-1]
    if n==m:
        print("Palidrome")
    else:
        print("NOt a Palindrome")
palindrome("madam")            
```

```python
#Menu driven
def square(n):
    return n*n
def cube(n):
    return n*n*n

choice=int(input("Enter 1.Square  2.Cube"))
n=int(input("Enter the number:"))
if choice==1:
    print(square(n))
elif choice==2:
    print(cube(n))
else:
    print("Invalid choice!")        


```

```python
#Anagram checking
def anagram(s1,s2):
    return sorted(s1.lower())==sorted(s2.lower())
print(anagram("listen","silent"))
```

```python
#Anagram removing the spaces and case
def anagram(s1,s2):
    s1=s1.replace(" ","").lower()
    s2=s2.replace(" ","").lower()
    return sorted(s1)==sorted(s2)
anagram("Dormitory","Dirty room")
```
