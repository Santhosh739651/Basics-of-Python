```python
#Local variables
def greet():
    msg="This is inside the function"
    print(msg)
    
greet()    
```

```python
#Global variables
msg="Hello"
def greet():
    print("Inside:",msg)
greet()
print("Outside:",msg)    
```

```python
#Accessing local variables
def fun():
    print("Inside Function", s)

s = "I love Geeksforgeeks"
fun()
print("Outside Function", s)
```

```python
#Modifying the Global variables
def fun():
    global s
    s+="SAnthosh"
    print(s)
s="Python"
fun() 
   
```
