```python
#Class--Blue print of the object
class Dog:
    def sound(self):
        print("Dog Barks!")
c1=Dog()     #Object
c1.sound()
```

```python
#Class with dataMembers
class Student:
    name="Santhosh"
    age=20
    def data(self):
        print("Name:",self.name)
        print("Age:",self.age)
s1=Student()    
s1.data()    
```

```python
#init--runs automatically when an object is created
class data:
    def __init__(self):
        print("Constructor is called without object!!")
c1=data()        
```

```python
#Intialize varaibles
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):                #Another method is created to store the data use for later
        print("Name:",self.name) 
        print("Age:",self.age)
c1=Student("Santhosh",20) 
c1.show()           
```

```python
#Multiple Objects
class names:
    def __init__(self,name):
        self.name=name
    def data(self):
        print(self.name)
c1=names("Santhosh")
c2=names("Python")
c1.data()
c2.data()            
```

```python
#Constructor with default value
class Employee:
    def __init__(self,name,salary=10000):
        self.name=name
        self.salary=salary
    def show(self):
        print("Name:",self.name,"Salary:",self.salary)
c1=Employee("Santhosh.")
c2=Employee("Python.",200000)
c1.show()
c2.show()            
```

```python
#Constructor with claculations
class Rectangle:
    def __init__(self,length,breadth):
        self.area=length*breadth
    def show(self):
        print("Area:",self.area)
r=Rectangle(10,5)
r.show()        
```

```python
#Use of super()--used to call the parentclass constructor from child class
class Parent:
    def __init__(self):
        print("Parent Constructor")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Constructor")
c=Child()                
```

```python
class Parent:
    def __init__(self):
        print("This is Parent")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child!!" )
c=Child()          
```
