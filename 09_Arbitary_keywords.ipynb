```python
#Arbitary keywords Arguments
# *args allows you to pass a varaible number of positional arguments to a function and a tuple
def abc(*values):
    for i in values:
        print(i)
abc("Santhosh","Python","Program")         
```

```python
#**kwargs used to allow the values as dictinaries
def func(**kwargs):
    for key,val in kwargs.items():
        print(key,val)
func(fruit="Mango")
```

```python
def func(**info):
    for key,value in info.items():
        print(f"{key}:{value}")
func(name="Santhosh",age=19,clas="AIDS")        
```

```python
#Combining *argv with extra arguments
def func(argv,*argv1):
    print("First arguments:",argv)
    for i in argv1:
        print("Arguments:",i)
func(12,13,14,"SAnthosh")        
```

```python
#Combining both
def combine(a,*args,**kwargs):
    print("a=",a)
    print("*args:",args)
    print("**kwargs:",kwargs)
combine(10,20,30303,"Hello",name="Santhosh")    
```

```python
#*argvs
def demo(*hi):
    print("Items:",hi)
demo(1,2,3,4)    
```

```python
#**kwargs
def sample(**Hello):
    print("Information:",Hello)
sample(Name="SANTHOSH",Age=19,Year="This year")    
```
