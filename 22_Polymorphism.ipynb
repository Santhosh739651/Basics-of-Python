```python
#Polymorphism->"many forms"
class Animal:
    def sound(self):
        print("Animal makes sound!")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meow")
a=Animal() 
b=Dog()       
c=Cat()
a.sound()
b.sound() 
c.sound()                       
```

```python
class pen:
    def write(self):
        return "Writing"
class pencil:
    def write(self):
        return "Not Writing"
d=pen()
e=pencil()
d.write()
e.write()        
```
