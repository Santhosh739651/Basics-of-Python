```python
#Inhertance--extracting features from parent class to child class
class Animal:
    def __init__(self,name):
        self.name=name
    def info(self):
        print("Animal name:",self.name)
class dog(Animal):
    def sound(self):
        print(self.name,"Barks")
d=dog("Jimmy")
d.info()
d.sound()                    
```

```python
#1.Single Inheritance
class Parent:
    def show(self):
        print("This is parent Class")
class child(Parent):
    def data(self):
        print("This is child class")
c=child()
c.show()
c.data()                
```

```python
class Person:
    def __init__(self,name,role):
        self.role=role
        self.name=name
class man(Person):        
    def show(self):
        print(self.name ,"is an ", self.role)
d=man("Santhosh","IAS")
d.show()
```

```python
#2.Multiple Inheritance
class parent:
    def data(self):
        print("This is Father class")
class mother:
    def data1(self):
        print("This is mother class")
class child(parent,mother):
    def data2(self):
        print("This is child")
c=child()
c.data()
c.data1()
c.data2()
```

```python
#3.Multilevel inheritance
class grandfather:
    def asthi(self):
        print("Thatha Asthi")
class father(grandfather):
    def asthi1(self):
        print("Nanna Asthi")        
class me(father):
    def noasthi(self):
        print("Naa asthi")
hi=me()
hi.asthi()
hi.asthi1()
hi.noasthi()        
```

```python
#4.Hierarchical
class Parent:
    def data(self):
        print("I am parent")
class child1(Parent):
    def show(self):
        print("I am Big son")
class child2(Parent):
    def show1(self):
        print("I am small son") 
c=child1()
c1=child2()
c.data()
c.show()
c1.data()
c1.show1()                           
```

```python
#5.Hybrid class
class A:
    def showA(self):
        print("Class A")
class B(A):
    def showB(self):
        print("Class B")
class C(A):
    def showC(self):
        print("Class C") 
class D(B,C):
    def showD(self):
        print("Class D")
c=D()
c.showA()
c.showB()
c.showC()
c.showD()                               
```
