```python
#Conditional statements--if,else,elif,match
num=0
if num <=0:
    print(num)

#if-else
n=int(input('Enter the number:'))
if n < 0:
    print("The number is negative..")

else:
    print("The number is positive")        

#if-elif-else statements
marks=int(input("Enter the marks of student:"))
if marks >=90:
    print("Grade:A")

elif marks >=80:
    print("Grade:B")   

elif marks >= 70:
    print("Grade : c")

else:
    print("Fail")             
```

```python
#Nested if statements
num=10
if num > 0:
    if num % 2 ==0:
        print("Number is positive even number")
    else:
        print("Number is positive odd number")

else:
    print("Number is negative")            
```

```python
#Leap year program
n=int(input("Enter the year:"))
if n %4 == 0:
    print("Leap year")
else:
    print("Not a leap year")    
```

```python
#Hhigh level programs on conditional statements
#Current Bill 
units=int(input("Enter the noof units consumed:"))
bill=0
if units<=100:
    bill=units*1.5
elif units<=200:
    bill=(units*1.5)+(units-100)*2.5
elif units<=300:
    bill=(units*1.5)+(units-200)*4+(100*2.5)
else:
    bill=(units*1.5)+(100*2.5)+(100*4)+(units-300)*6

print("Electricity Bill:$",bill)           
```

```python
#Withdrawl amount in ATM
Balance=100000
amount=int(input("Enter the amount to withdrawl:"))
if amount<=0:
    print("Invalid Amount")
elif amount>Balance:
    print("In-sufficient funds")
elif amount%100!=0:
    print("Please enter the amount in multiples of 100")
else:
    Balance-=amount
    print("Withdrawl money Successfully")
    print("Balance amount:",Balance)            
```

```python
#match case--same as the switch case
day=int(input("Enter Day number (1-7):"))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Daynumber")
                                    


```

```python
#Calculator
a=int(input("Enter the a value:"))
b=int(input("Enter the b value:"))
op=input("Enter th operation-(+,-,*,/):")
match op:
    case "+":
        print("Result:",a+b)
    case "-":
        print("Result:",a-b)
    case "*":
        print("Result:",a*b)
    case "/":
        print("Result:",a/b)   
    case _:
        print("Invalid Operator")             
```

```python
#Vowel or Consonant
ch=input("Enter the Letter:")
match ch:
    case "a" | "e"|"i"|"o"|"u":
        print("Vowels")
    case _:
        print("Consonants")    
```
