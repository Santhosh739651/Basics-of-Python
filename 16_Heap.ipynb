```python
#Heapq--smallest element is always at 0 index
import heapq
lst=[10,20,30,5,89]
heapq.heapify(lst)
print(lst)
```

```python
#Insert element into a heap
import heapq
data=[10,20,30,40]
heapq.heappush(data,9)
print(data)
```

```python
#Remove element  --smallest element is popped
import heapq
inform=[1,2,4,5,3,6]
x=heapq.heappop(inform)
print("Popped element:",x)
print("Heap:",inform)
```

```python
#Push and Pop together
import heapq
matter=[1,2,3,4,5,6,7]
x=heapq.heappushpop(matter,7)
print(x)
print(matter)
```

```python
#Greatest and Smallest 
import heapq
data=[123,4566,676,34677,356]
print("Smallest:",heapq.nsmallest(1,data))
print("Largest:",heapq.nlargest(1,data))
```
