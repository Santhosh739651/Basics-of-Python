```python
#Loops--for,while,break,continue,pass
#for loop
#printing of 1 to 20 odd numbers 
for i in range (1,5,2):
    print(i)

#Sum of n natural numbers
n=int(input("Enter the num to do sum:"))
total=0
for i in range(1,n+1):
    total+=i
print("Total:",total) 

#Printing square of a number
start = int(input("Enter the starting range:"))
end = int(input("Enter the end range:"))
for i in range(start,end):
    print(i,"Square = ",i*i)
```

```python
#For loop tough problems
#Reverse a Number without slicing
n=int(input("enter the number:"))
rev=0
for i in range(len(str(n))):
    digit=n%10
    rev=rev*10+digit
    n//=10

print(rev)     
```

```python
#While loop         //we should initialize i value
n=int(input("Enter number:"))
i=1
while i<=n:
    print(i)
    i+=1

#Sum on N natural numbers
n=int(input("Enter Number:"))
i=1
sum=0
while i<=n:
    sum+=i
    i+=1
print("Sum:",sum)

```

```python
#Armstrong  Number
n=int(input("Enter the number:"))
temp=n
sum=0
digit=len(str(n))
while n>0:
    sum+=(n%10)**digit
    n//=10

if sum==temp:
    print("Armstrong Number")
else:
    print("Not an Armstrong")        
```

```python
#Prime numbers
n=int(input("Enter the number:"))
i=2
flag=1
if n<=0:
    flag=0

while i<=n//2:
    if n%i==0:
        flag=0
        break
    i+=1
    if flag==1:
        print("Prime numbers")

    else:
        print("Not a prime")    


```

```python
#Skipping negative number
while True:
    n=int(input("Enter the number:"))
    if n<0:
        continue
    if n==0:
        break
    print("Positive number",n)
```

```python
#Skipping vowels in a string
word=input("Enter the String:")
vowels="aeiouAEIOU"
for ch in word:
    if ch in vowels:
        continue
    print(ch,end=" ")
```
