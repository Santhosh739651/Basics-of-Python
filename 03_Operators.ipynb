```python
#1.Arithmetic Operators--+,-,*,/,//,%,**
a=20
b=5
print("Arithmetic Operators:")

print("\nAddition:",a+b)
print("Subtraction:",a-b)
print("Multiplication:",a*b)
print("Division:",a/b)
print("Floor Division:",a//b)
print("Modulus:",a%b)
print("Exponentation:",a**b)
```

```python
#2.Comparision Operators--==,<,>=,<=,!=
a=100
bin=1000
print("Comparison Operators:")
print("\n",a==b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
print(a!=b)
```

```python
#3.Logical Operators--and,or,not
a=True
b=False
print("Logical Operators:")
print(a and b)
print(a or b)
print(not b)
```

```python
#4.Bitwise Operators--&,|,^,~,<<,>>
a=5 # 0101 in binary
b=3 # 0011 in binary
print("\n Bitwise Operators:")
print("a&b:",a&b)                #Bit wise And-- 1&1=1 else 0.shows output in number which directly converts from binary
print("a|b:",a|b)                #Bitwise OR--   Any bit is 1 then 1 else 0.
print("a^b:",a^b)                #Bitwise XOR--same bits->1,else->0
print("~a:",~a)                  #Bitwise Not
print("a<<1:",a<<1)              #Shifts move to left by multiply by (x<<n=x*2**n)
print("a>>1:",a>>1)              #x>>n=x%2**n

```

```python
#5.Assingemnt operators--=,+=,-=,*=,/=,%=,**=
a=10
a+=5
print("Assingment Opeartors:")
print("\n a+=5:",a)

a-=2
print("a-=2:",a)

a*=2
print("a*=2:",a)

a/=2
print("a/=2:",a)

a%=2
print("a%=2:",a)

a**=2
print("a**=2:",a)
```

```python
#6.Identity Operators--is,is not
a=10
b=20
print("Identity Operator:")
print(" a is b",a is b)
print(" a is not b:",a is not b)
```

```python
#Membership operator--in,not in
t=(1,2,3,4)
print(2 in t)
print(100 not in t)
```
