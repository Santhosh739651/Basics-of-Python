```python
#Functools--built in modules that provide higher order functions..
#1.Reduce
from functools import reduce
lst=[1,2,3,4,5]
result=reduce(lambda x,y:x+y,lst)
print(result)
```

```python
#2.Partial
from functools import partial
def fine(a,b):
    return a**b
power=partial(fine,a=3)
print(fine(2,4))
```

```python
#3.Partialmethod class--Creates new functon
from functools import partialmethod
class Demo:
    def __init__(self):
        self.color='black'
    def _color(self,type):
        self.color=type
    set_red=partialmethod(_color,type='red')
obj=Demo()
print(obj.color)  
obj.set_red()
print(obj.color)      
```
