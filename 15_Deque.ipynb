```python
#Deques--used for fast insertion and deletion from both ends
#creating deque
from collections import deque
dq=deque([10,20,30,40])
print(dq)
print(type(dq))
```

```python
#Adding elements at both end
from collections import deque
dq=deque()
dq.append(10)
dq.append(20)
dq.appendleft(5)
print(dq)
```

```python
#Removing the elements
from collections import deque
dq=deque([1,2,3,4,5])
dq.pop()
print(dq)
dq.popleft()
print(dq)
```

```python
#Reversing the Deque
from collections import deque
dq=deque([1,2,3,4,5])
dq.reverse()
print(dq)
```

```python
#Rotating
from collections import deque
dq=deque([1,2,3,4,5])
print(dq)
dq.rotate(1)
print(dq)
dq.rotate(2)
print(dq)
dq.rotate(-2)
print(dq)
dq.rotate(-1)
print(dq)
```

```python
#Queue using deque
from collections import deque
queue=deque()
queue.append("A")
queue.append("B")
queue.append("C")
print(queue.popleft())
print(queue.popleft())
print(queue)
```

```python
#Stacks using Deque
from collections import deque
stack=deque()
stack.append(10)
stack.append(20)
stack.append(30)
print(stack)
stack.pop()
print(stack)
```
