```python
def Squares():
    n=1

    while n<=10:
        sqr=n*n
        yield sqr
        n+=1

values=Squares()

for i in values:
    print(i)
```

```python
def count(n):
    for i in range(1,n+1):
        yield i


for value in count(5):
    print(5)        
```
