```python
#Iterator -- allows to traverse elements one by one
#It is an object
#__iter__():Returns the iterator object itself
#__next___():Returns next value from sequence
#yeild(): creates iterator automatically

numbers=[12,3,4,5,6]
it=iter(numbers)
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

```python
#Internally for loop uses iterator
names=["santhosh","Python","Program"]
for name in names:
    print(name)
```

```python
def even_numbers(n):
    for i in range(2,n+1,2):
        yield i
c=even_numbers(10)
for num in even_numbers(10):
    print(num)
```
