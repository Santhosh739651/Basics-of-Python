```python
#Read file line by line 
file=open("sample.txt","r")
line1=file.read()
line2=file.read()
print(line1)
```

```python
#Reading all linesin a file
file=open("sample.txt","r")
lines=file.readlines()
file.close()

print("Lines in the file :")
for line in lines:
    print(line)
```

```python
#Appending data in to a file
file=open("sample.txt","a")
file.write("\n this line is appended successfully...")
file.close()
print("Appended is done..")
```

```python
#Using with
with open("sample.txt","r") as file:
    data=file.read()

print(data)    
```

```python
#Copying contents from one file to another file

with open("sample.txt","r")as file:
    data=file.read()

with open("copy.txt","w")as anotherfile:
    anotherfile.write(data)

print("The data is copied successfully...")          
```

```python
#Counting numbers of a line
count=0
with open("sample.txt","r")as file:
    for line in file:
        count+=1

print("Number of lines:",count)        
```

```python
#Deleting a file

import os
os.remove("copy.txt")
print("The file is deleted successfully..")
```
