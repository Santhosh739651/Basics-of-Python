```python
#Factorial of a number
def factorial(n):
    if n==1 or n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(5))    
```

```python
#Fibonacci series
def series(n):
    a,b=0,1
    for i in range(n):
        print(a,end=" ")
        a,b=b,a+b
series(10)        
        
```

```python
#Sum of n natural numbers
def natural(n):
    if n==0 or n==1:
        return 1
    else:
        return n+natural(n-1)
print(natural(4))    
```

```python
#Power of a number
def power(x,n):
    if n==0:
        return 1
    return x*power(x,n-1)
print(power(2,5))
```

```python
#Sum of digits of a number
def num(n):
    if n ==0:
        return 0
    return (n % 10)+num(n//10)
print(num(153))
```
