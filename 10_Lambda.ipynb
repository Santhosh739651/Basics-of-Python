```python
#Lambda function is used to short or small problems
s1="santhosh"
s2=lambda x:x.upper()
print(s2(s1))
```

```python
#Checking positive or negative

n=lambda x:"Positive" if x>0 else "Negative" 
print(n(-3))
```

```python
#Printing square and cube of number
sq=lambda x:x**2
print(sq(6))

cube=lambda x:x**3
print(cube(3))
```

```python
li=[lambda arg=x:arg*10 for x in range(1,6)]
for i in li:
    print(i())
```

```python
#Lambda with multiple statements
calc=lambda x,y :(x+y,x*y)
print(calc(5,6))
```

```python
#Bult-in functions--map(),filter(),reduce()
#map()-performs the same operation on all elements
numbers=[1,2,3,4,5]
squares=list(map(lambda x:x*x,numbers))
print(squares)
```

```python
#Conerting strings to integers
nums=["10","20","30","40"]
result=list(map(int,nums))
print(result)
```

```python
#Adding five to each element
nums=[10,20,30,40,50]
def add(x):
    return x+5
result=list(map(add,nums))
print(result)
```

```python
#Finding the length of each string
words=["Santhosh","Python","Programming"]
result=list(map(len,words))
print(result)
```

```python
#Adding to list using map
a=[10,20,30,40]
b=[50,50,60]
result=list(map(lambda x,y:x+y,a,b))
print(result)
```

```python
#Filter--selects the from an iterable which satisfy the condition
numbers=[10,20,34,59606,3,11]
result=list(filter(lambda x:x%2==0,numbers))
print(result)
```

```python
#Reduce--reduces the values to single unit
from functools import reduce
numbers=[10,20,30]
result=reduce(lambda a,b:a+b,numbers)
print(result)
```

```python
#Filter with starting with a vowel
names=["Santhosh","orange","ant","on"]
vowels=["a","e","i","o","u"]
result=list(filter(lambda x:x[0].lower() in vowels,names))
print(result)
```

```python
#Maximum using reduce
from functools import reduce
nums=[10,203,4050,50]
result=reduce(lambda x,y:x if x>y else y,nums)
print(result)
```
