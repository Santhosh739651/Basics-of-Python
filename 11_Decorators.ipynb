```python
#Modify the function without modifying another function
def decorator(func):
    def wrapper():
        print("Before Execution..")
        func()
        print("After the execution..")
    return wrapper
    
@decorator
def hello():
    print("Hello world")

hello()    
```

```python
def greet_function(func):
    def jam():
        print("Hello!")
        func()
        print("Pyhton")
    return jam
@greet_function
def hi():
    print("Welcome To")
hi()       
```

```python
#Decorator with arguments
def details(func):
    def data(name):
        print("Hello")
        func(name)
        print("Thank you")
    return data 
@details
def show(name):
    print(name)
show("Santhosh")        
```

```python
#Decorator with *args,**kwargs
def decorator(func):
    def wrapper(*args,**kwargs):
        print("Before Performing")
        func(*args,**kwargs)
        print("After execution ")
    return wrapper
@decorator
def cls(a,b):
    print("Sum:",a+b)  
cls(10,20)      
```

```python
#Multiple Decorators
def decor1(func):
    def wrapper():
        print("Starting of Decor-1")
        func()
        print("Ending of Decor-1")
    return wrapper
def decor2(func):
    def wrapper():
        print("Starting of Decor-2") 
        func()
        print("Ending of decor-2")
    return wrapper
@decor1
@decor2
def show():
    print("Hello!!") 
show()          
```
