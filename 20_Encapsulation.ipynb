```python
#Encapsulation--Hiding important data+controlled access
#It is represented by "_",ex:-_name
#Data cannot be accesed directly
#get -- used to access data
#Set -- used to modify data
```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self._salary=salary
emp=employee("Santhosh",50000)
print(emp.name)     
print(emp._salary)  #We have call it seperately to show data
```

```python
#Accessed with in same class and sub class only
class employee:
    def __init__(self,name,age):
        self.name=name
        self._age=age
class sub(employee):        
    def show_age(self):
        print("Age:",self._age) 
emp=sub("Santhosh",21)
print(emp.name) 
emp.show_age()       

```

```python
class employee:
    def __init__(self,name,salary):
        self.name=name
        self._salary=salary
    def show(self):
        print("Salary:",self._salary)
c=employee("Santhosh",50000) 
c.show()           
print(c.name)        
```

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self._age=age
    def get_age(self):
        return self._age
    def set_age(self,age):
        self._age=age
p=person("Santhosh",20)
print(p.get_age())
p.set_age(22)
print(p.get_age())           
```

```python
#Bank account
class BankAccount:
    def __init__(self,balance):
        self._balance=balance
    def deposit(self,amount):
        self._balance+=amount
        print("Deposited:",amount)
    def withdrawl(self,amount):
        self._balance-=amount
        print("Withdrawl:",amount)
    def get_balance(self):
        return self._balance
s=BankAccount(100)
s.deposit(50)
s.withdrawl(100)
print("Balance:",s.get_balance())            
```

```python
#Student marks
class student:
    def __init__(self,marks):
        self._marks=marks
    def set_marks(self,marks):
        if 0<=self._marks>=100:
            self._marks=marks
        else:
            print("Invalid Marks!")
    def get_marks(self):
        return self._marks
s=student(56) 
s.get_marks() 
s.set_marks(150)
print("Result:",s.get_marks())            
```
