```python
#DEfaultdict--Automatically gives a value when key does not exist
from collections import defaultdict
d=defaultdict(int)
d["a"]=d["a"]+1
d["b"]=d["b"]+1
d["a"]=d["a"]+1
print(d)
```

```python
#Count words
from collections import Counter
s="apple banana apple orange banana apple"
word=s.split()
d=defaultdict(int)
for w in word:
    d[w]+=1
print(d)
```

```python
#Storing of values in lists
d=defaultdict(list)
d["a"].append(20)
d["a"].append(10)
d["b"].append(10)
print(d)
```

```python
#Using the lambda function
d=defaultdict(lambda : "NOt present")
d["a"]=1
d["b"]=2
print(d["a"])
print(d["b"])
print(d["c"])
```

```python
d=defaultdict(list)
for i in range(5):
    d[i].append(i)
print(d)
```

```python
sd=defaultdict(str)
sd["Greeting"]="Hello"
print(sd)
```
