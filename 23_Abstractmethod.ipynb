```python
#Abstract mthod--NO body of implementation
#--It only tells what to do,not how to do,
#we should import the abc(abstract base class)

from abc import ABC,abstractmethod
class computer(ABC):
    @abstractmethod
    def process(self):
        pass
class Laptop(computer):
    def process(self):
        print("Its runnig..")  
c=Laptop()
c.process()          
```

```python
from abc import ABC,abstractmethod
class company(ABC):
    def __init__(self,name):
        self.name=name
    @abstractmethod
    def model(self):
        pass
class type(company):
    def model(self):
        print(self.name,"has two wheels!!")
c=type("JAVA")
c.model()                
```

```python
from abc import ABC,abstractmethod
class Area:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    @abstractmethod
    def show(self):
        pass 
class Rectangle(Area):
    def show(self):
        print("Area of Rectangle:",self.breadth * self.length)
ans=Rectangle(10,3)
ans.show()        
```

```python
from abc import ABC,abstractmethod
class Vehicle:
    @abstractmethod
    def key(self):
        pass
class Car(Vehicle):
    def key(self):
        print("Car Starts with Button")
class Bike(Vehicle):
    def key(self):
        print("Bike starts with keys")
c=Car()
b=Bike()
c.key()
b.key()                    
```

```python
from abc import ABC,abstractmethod
class Employee:
    def __init__(self,name):
        self.name=name
    @abstractmethod
    def details(self):
        pass
class Emp(Employee):
    def details(self):
        return self.name 
c=Emp("Santhosh")
c.details()            
```
