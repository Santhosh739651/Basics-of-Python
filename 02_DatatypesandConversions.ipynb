```python
#1.Numeric-> int,float,complex
#int
a = 10
print("Data type of a:",type(a))

#Float
b=8.4
print("Data type:",type(b))

#Complex
c=3+5j
print("Data type:",type(c))


```

```python
#2.Sequence datatype->string,list,tuple
#Strings(immutable)
name='Santhosh'
print("Data type:",type(name))

#List(mutable)
detalis=['Santhosh',19,'Aids-2']
print("Data type:",type(detalis))

#tuple(immutable)
data=("Santhosh",10,'Aids-2')
print("Data type:",type(data))
```

```python
#String operations
s = "Hello python"
print(s[2])             #printing using index numbers
print(s[-1])
print(s[0:5])
print(s[6:13])
print(s[0:13:2])         #[ start : stop : step ]
print(s[::-1])

#String Concatination
a="Annagiri"
b="Santhosh"
print(a+" "+b)

#String repetetion
num="Hi"
print(num * 3)

#Built-in string functions
print(len(num))
print(max(s))
print(min(num))

#String methods
print(num.lower())
print(num.upper())
print(num.title())
print(num.find("i"))     #searching the index number

#Replacing the strings
abc="I do Python"
print(abc.replace("Python","JAVA"))

#Joining the strings
words=["Python","is","Easy"]
print(" ".join(words))

#Formatting strings
name="santhosh"
age=20
print(f"My name is {name} and my age is {age}.")
```

```python
#List operations
numbers=[1,2,3,4,5]
print("list of numbers:",numbers)
fruits=["Mango","Guava","Grapes"]
print("Fruits:",fruits)

#Accessing elements
print(fruits[-1])        #use index numbers to specify
print(fruits[1])

#Appending an element
fruits.append("Banana")
print(fruits)

#Inserting an element
fruits.insert(1,"Kiwi")   #We use index numbers to insert the numbers
print(fruits)

#Removing the element
fruits.remove("Mango")
print(fruits)

#Updating a value
fruits[2]="Water Melon"
print(fruits)

#Length of the list
print(len(fruits))

#Sorting of numbers
num=[20,356,274,89]
num.sort()
print(num)
print(num[::-1])         #reversing a list

a=[10,20]
b=[30,40]
c=a+b
print(c)
```

```python
#Tuple operations-(immutables)
t=(1,2,3,4)
print("Data type:",type(t))
print(t)

#Tuple slicing
t=(10,20,30,40)
print(t[1])
print(t[::-1])

#tuple concatention
a=(1,2)
b=(3,4)
c=(5,6)
print(a+b+c)

#Printing the repetions
x=(10,20)
print(x*3)

#Membership operation
y=(10,20,30)
print(10 in y)
print(50 in y)

#Using the iterations
xyz=(99,103,385,17944)
for i in xyz:
    print(i)

#Using the built-in functions
t=(10,20,30,40,50)
print(t.count(20))         #count is used find index number,#index is used find index number
print(t.index(40))

#Converting to list
lst=[10,20,30]
t=tuple(lst)
print(t)
```

```python
#3.Dictionaries(key,value pair)
Data={"name":"Santhosh","Age":19,"Course":"Python"}
print("Data type:",type(Data))
print(Data)

#Operations on Dictionaries
print(Data['name'])
print(Data.get("Course"))
print(Data["Age"])

#Adding new elements
Data["Grade"]="A"

#Updating Data
Data["Age"]=20

#Removing elements
del Data["Grade"]
Data.pop("name")

#Dictinary methods
print(Data.keys())         #prints all keys in dictionary
print(Data.values())       #prints all values in dictinary
print(Data.items())        #prints all items in dictionary

#Checking existence of key 
if "Age" in Data:
    print("The key is existed")

#Length of dictionary
print(len(Data))  

#Merging 
dict1={"a":1,"b":2}
dict2={"c":3}
dict1.update(dict2)
print(dict1)
```
