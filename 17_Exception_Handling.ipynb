```python
try:
    a=int(input("Enter the a value:"))
    b=int(input("Enter the b value:"))
    result=a/b
    print("Result:",result)
except Exception as e: #Exception is used to catch all the exceptions in the program
    print("Division by zero is not possible..")    
```

```python
try:
    n=int(input("Enter the values :"))
    print("Entered values:",n)
except ValueError:
    print("ERROR..Please enter the values in integer ")    
```

```python
try:
    a=int(input("Enter the a value:"))
    b=int(input("Enter the value of b:"))
    division=a/b
    print(division)
except ZeroDivisionError:
    print("Division by zero is not possible")
except ValueError:
    print("Enter the int values only..")    

```

```python
#Using else bloock
try:
    num1=int(input("Enter the values of num1:"))
    num2=int(input("Enter the values of num2:"))
    c=num1+num2
except ValueError:
    print("Invalid value")
else:
    print(c)    

```

```python
#Using finally
try:
    n=int(input("enter a number:"))
    print(n*n)
except Exception:
    print("Enter the value input")
finally:
    print("The execution of the program is completed")        
```
