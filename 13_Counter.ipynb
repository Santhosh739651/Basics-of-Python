```python
#Counter--Shows how many times value appears
from collections import Counter
val=[1,2,2,3,4,4,4,3,1]
val1=["a","b","c","d","a","b"]
print(Counter(val))
print(Counter(val1))
```

```python
#Creating the counter
ctr1=Counter([1,2,3,4,4,3,2,1,4,233])
print(ctr1)
```

```python
#Updating counters
ctr1=Counter([1,2,3,4,4,3,2,1,3])
ctr1.update([9,8,7])
print(ctr1)
```

```python
#counter Methods
#1.elements()
ctr=Counter([1,2,3,3,4,5,2,5])
items=list(ctr.elements())
print(items)
```

```python
#2.most_common()
ctr=Counter([1,2,3,3,4,5,2,6,6,6])
co=ctr.most_common()
print(co)
```

```python
#3.Subtract
ctr=Counter([1,2,2,3,4,])
ctr.subtract([1,4])
print(ctr)
```

```python
#Find the least common Element
from collections import Counter
lst=Counter([1,2,2,3,3,4,4,5,5])
least=lst.most_common()[-1]
print(least)
```

```python
#Counting the words
from collections import Counter
s="python is easy and python is useful"
word=s.split()
c=Counter(word)
print(c)
```
