```python
class StringIterator:
    def __init__(self, text):
        self.text = text
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration

s = StringIterator("PYTHON")

for c in s:
    print(c)
```

```python
class numbers:
    def __init__(self,n):
        self.n=n
        self.number=1
    def __iter__(self):
        return self
    def __next__(self):
        if self.number<self.n:
            value=self.number
            self.number+=1
            return value
        else:
            raise StopIteration

D=numbers(10)

for i in D:
    print(i)
```

```python
#Even numbers
class Even:
    def __init__(self,n):
        self.n=n
        self.number=2
    def __iter__(self):
        return self
    def __next__(self):
        if self.number<self.n:
            value=self.number
            self.number+=2
            return value
        else:
            raise StopIteration

values=Even(10)

for i in values:
    print(i)
```
