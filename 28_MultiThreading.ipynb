```python
#MultiThreading--It allows a program to run multiple tasks at the same time using threds..
from time import sleep
from threading import*
class Hello(Thread):
    def run(self):                               #run()-> is used like a function name
        for i in range(5):
            print("Hello")
            sleep(1)  
class Hi(Thread):  
    def run(self):
        for i in range(5):
            print("Hi")
            sleep(1)
t1=Hello()
t2=Hi()
t1.start()                                       #start()->used to call the class
sleep(0.2)
t2.start()  

t1.join()
t2.join()
print("Bye")
```

```python
from threading import*
from time import sleep
class Mythread(Thread):
    def run(self):
        for i in range(1,3):
            print("Thread is Running:",i)
            sleep(1)
t1=Mythread()
t2=Mythread()
t3=Mythread()
t1.start()
t2.start()
t3.start()
t1.join()
print("Thread is ended!!")

        
```

```python
#Different tasks
from threading import*
from time import sleep    
class Number(Thread):
    def run(self):
        for i in range(1,6):
            print("Number: ",i)
            sleep(1)
class Letters(Thread):
    def run(self):
        abc="ABCDE"
        for ch in abc:
            print("Letters:",ch)
            sleep(1)
t1=Number()
t2=Letters()
t1.start()
sleep(0.2)
t2.start()                        
```
